```python
import pandas as pd
import numpy as np
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
data = [
    ['Milk', 'Bread', 'Butter'],
    ['Bread', 'Butter'],
    ['Milk', 'Bread'],
    ['Milk', 'Bread', 'Butter', 'Eggs'],
    ['Bread', 'Eggs'],
    ['Milk', 'Eggs'],
    ['Milk', 'Bread', 'Eggs'],
    ['Bread', 'Butter', 'Eggs'],
    ['Milk', 'Butter'],
    ['Milk', 'Bread', 'Butter', 'Eggs']
]

transactions = pd.DataFrame(data)
transactions
```

```python
items = sorted(set(item for transaction in data for item in transaction))

basket = pd.DataFrame(0, index=range(len(data)), columns=items)

for i, transaction in enumerate(data):
    for item in transaction:
        basket.loc[i, item] = 1

basket
```

```python
frequent_itemsets = apriori(
    basket,
    min_support=0.3,
    use_colnames=True
)

frequent_itemsets
```

```python
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.5
)

rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']]
```

```python
rules = rules.sort_values(
    by='confidence',
    ascending=False
)

rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']]
```

```python
import pandas as pd
from collections import defaultdict

students = [
    ("Rahul", "Python", "A"),
    ("Priya", "Python", "S"),
    ("Aman", "Python", "B"),
    ("Neha", "Python", "A"),
    ("Riya", "Python", "C"),
    ("Karan", "Python", "B"),
    ("Rahul", "DBMS", "B"),
    ("Priya", "DBMS", "A"),
    ("Aman", "DBMS", "C"),
    ("Neha", "DBMS", "S"),
    ("Riya", "DBMS", "B"),
    ("Karan", "DBMS", "D"),
    ("Rahul", "Big Data", "A"),
    ("Priya", "Big Data", "S"),
    ("Aman", "Big Data", "B"),
    ("Neha", "Big Data", "A"),
    ("Riya", "Big Data", "C"),
    ("Karan", "Big Data", "B")
]

def mapper(data):
    mapped_data = []
    for student, subject, grade in data:
        mapped_data.append(((subject, grade), 1))
    return mapped_data

def reducer(mapped_data):
    result = defaultdict(int)
    for key, value in mapped_data:
        result[key] += value
    return result

mapped = mapper(students)
reduced = reducer(mapped)

result = pd.DataFrame(
    [
        [subject, grade, count]
        for (subject, grade), count in reduced.items()
    ],
    columns=["Subject", "Grade", "Number of Students"]
)

pivot_result = result.pivot(
    index="Subject",
    columns="Grade",
    values="Number of Students"
).fillna(0).astype(int)

pivot_result
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import sum
```

```python
spark = SparkSession.builder \
    .appName("Sales Analysis") \
    .getOrCreate()

print("Spark Session Created Successfully")
```

```python
sales_data = [
    ("Laptop", "Electronics", 5),
    ("Mobile", "Electronics", 10),
    ("Headphones", "Electronics", 15),
    ("Shirt", "Clothing", 20),
    ("Jeans", "Clothing", 12),
    ("Shoes", "Clothing", 8),
    ("Rice", "Grocery", 25),
    ("Wheat", "Grocery", 30),
    ("Oil", "Grocery", 18),
    ("Notebook", "Stationery", 15),
    ("Pen", "Stationery", 40),
    ("Pencil", "Stationery", 35)
]

columns = ["Product", "Category", "Quantity"]

sales_df = spark.createDataFrame(
    sales_data,
    columns
)

sales_df.show()
```

```python
category_sales = sales_df.groupBy("Category") \
    .agg(sum("Quantity").alias("Total Products Sold"))

category_sales.show()
```

```python
category_sales \
    .orderBy("Category") \
    .show()
```

```python
result_pandas = category_sales.toPandas()

result_pandas
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.bar(
    result_pandas["Category"],
    result_pandas["Total Products Sold"]
)

plt.xlabel("Product Category")
plt.ylabel("Number of Products Sold")
plt.title("Number of Products Sold in Each Category")

plt.show()
```
